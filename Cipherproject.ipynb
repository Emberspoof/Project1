```python
def secret(message):
    dictciph = {" " : " ", "n" : "a", "q" : "b", "o" : "c", "u" : "d", "i" : "e", "w" : "f", "x" : "g", "d" : "h", "f" : "i", "k" : "j", "l" : "k", "s" : "m", "e" : "n", "g" : "o", "z" : "p", "v" : "q", "t" : "r", "b" : "s", "h" : "t", "r" : "u", "c" : "v", "m" : "w", "p" : "x", "j" : "y", "y" : "z"}
    ciphered = ""
    for letter in message:
        for key, value in dictciph.items():
            if letter == value:
                ciphered += key
                
    return ciphered

message_1 = "nicholas is from louisana and loves frogging"
message_2 = "jack has tiger as his middle name"
message_3 = "ember likes to build their own computers"

print (secret(message_1))
print (secret(message_2))
print (secret(message_3))
```
